```python
query = """
SELECT
  sales.sales_id,
  sales.order_id,
  customer.customer_key,
  products.product_key,
  sales.region,
  order_date_dim.date_key as order_date_key,
  ship_date_dim.date_key as ship_date_key,
  sales.order_date,
  sales.ship_date,
  sales.ship_mode,
  sales.sales,
  sales.discount,
  sales.quantity,
  sales.profit,
  -- derived columns
  date_diff(sales.ship_date, sales.order_date) as delivery_days,
  sales.sales * sales.discount as discount_amount ,
  sales.sales / (1 - sales.discount) as gross_sales


FROM global_mart_retail_dev.silver.sales AS sales
LEFT JOIN global_mart_retail_dev.gold.dim_customer AS customer
  ON customer.customer_id = sales.customer_id
LEFT JOIN global_mart_retail_dev.gold.dim_products AS products
  ON products.product_id = sales.product_id
LEFT JOIN global_mart_retail_dev.gold.dim_date AS order_date_dim
  ON order_date_dim.full_date = sales.order_date
LEFT JOIN global_mart_retail_dev.gold.dim_date AS ship_date_dim
  ON ship_date_dim.full_date = sales.ship_date


"""
df = spark.sql(query)
```

```python
display(df)
```

```python
df.write.mode("overwrite").format("delta").saveAsTable("global_mart_retail_dev.gold.fact_sales")
```
