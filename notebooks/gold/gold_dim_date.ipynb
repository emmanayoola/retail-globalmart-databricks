```python
query = """
SELECT
  cast(date_format(d, 'yyyyMMdd') as int) as date_key,
  d as full_date,
  year(d) as year,
  quarter(d) as quarter,
  concat('Q', quarter(d)) as quarter_name,
  month(d) as month,
  date_format(d, 'MMMM') as month_name,
  date_format(d, 'MMM') as month_name_short,
  cast(date_format(d, 'yyyyMM') as int) as month_year_sort,
  date_format(d, 'MMM yyyy') as month_year,
  concat('Q', quarter(d), ' ', year(d)) as quarter_year,
  weekofyear(d) as week_of_year,
  dayofmonth(d) as day_of_month,
  dayofweek(d) as day_of_week,
  date_format(d, 'EEEE') as day_name,
  case when dayofweek(d) in (1, 7) then 1 else 0 end as is_weekend,
  cast(date_format(d, 'yyyy-MM-dd') as date) as date_string
FROM (
    SELECT explode(sequence(to_date('2014-01-01'), to_date('2018-12-31'), interval 1 day)) as d)
"""
df = spark.sql(query)
```
