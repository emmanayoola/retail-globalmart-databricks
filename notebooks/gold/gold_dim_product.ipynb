```python
query = """
SELECT
  ROW_NUMBER() OVER (ORDER BY p.product_id) AS product_key,
  p.product_id,
  p.product_name,
  p.category,
  p.sub_category,
  p.product_hash,
  p.valid_from,
  p.valid_to,
  p.is_current_flag,
  p.load_timestamp
FROM global_mart_retail_dev.silver.products AS p
"""
df = spark.sql(query)
```

```python
df.limit(10).display()
```

```python
df.write.mode("overwrite").format("delta").saveAsTable("global_mart_retail_dev.gold.dim_products")
```
